```python
#count instances Per Class

from collections import Counter

class_names = ['carton box', 'cracked carton box', 'opened carton box', 'wet carton box']

def count_classes_in_split(labels_dir):
    class_counts = Counter()
    total_label_files = 0

    for filename in os.listdir(labels_dir):
        if not filename.endswith('.txt'):
            continue
        total_label_files += 1
        filepath = os.path.join(labels_dir, filename)
        with open(filepath, 'r') as f:
            for line in f:
                parts = line.strip().split()
                if parts:
                    class_idx = int(parts[0])
                    class_counts[class_idx] += 1

    return class_counts, total_label_files

for split in ['train', 'valid', 'test']:
    labels_path = f'{split}/labels'
    counts, num_files = count_classes_in_split(labels_path)

    print(f"\n{'='*50}")
    print(f"{split.upper()} — {num_files} label files")
    print(f"{'='*50}")
    for idx, name in enumerate(class_names):
        print(f"  [{idx}] {name}: {counts.get(idx, 0)} instances")
```
